seaborn Kdeplot


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("tips")
sns.histplot(data=df, x="tip", kde=True)
plt.show()
```

```python
df = sns.load_dataset("tips")
sns.rugplot(data=df, x="tip")
plt.show()
```

```python
df=sns.load_dataset("tips")
sns.kdeplot(data=df, x="tip")
plt.show()
```

seaborn *category*

```python
sns.countplot(data=df, x="day", hue="sex")
```

```python
ax=sns.barplot(data=df, x="day", y="sex", estimator=np.sum)
ax.set_xlabel("axes x label")
```

seaborn boxplot

```python

sns.boxplot(data=df, x="day", y="total_bill")
plt.show()
```

```python
sns.set_style("darkgrid")

sns.boxplot(data=df, x="day", y="total_bill", palette="pastel")
plt.show()
```

Heat map are pivot

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")

corr = data.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

flights = sns.load_dataset("flights")

pivot = flights.pivot_table(
    index="month",
    columns="year",
    values="passengers"
)

sns.heatmap(pivot, annot=True, fmt=".0f", cmap="coolwarm")

plt.show()
```

```python
sns.pairplot(data=data)
```

```python
grid = sns.PairGrid(
   data=data

)

grid.map_diag(sns.histplot)
grid.map_lower(sns.kdeplot, fill=True)
grid.map_upper(sns.kdeplot, fill=True)

plt.show()
```

```python
grid = sns.PairGrid(
    data=data,
    hue="sex"
)

grid.map_diag(sns.histplot)
grid.map_lower(sns.kdeplot, fill=True)
grid.map_upper(sns.kdeplot, fill=True)

plt.show()
```

```python
grid=sns.FacetGrid(
    data=data,
    row="sex",
    col="time"
)
grid.map(sns.scatterplot, "total_bill", "sex")
```
