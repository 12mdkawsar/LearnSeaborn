```python
data = sns.load_dataset("brain_networks")
# The first 3 rows are network, node, and hemi names. We slice from index 3 onwards and convert to float.
numeric_data = data.iloc[3:, :5].astype(float)
sns.histplot(data=numeric_data, kde=True)
```

```python

# Titanic dataset load
titanic = sns.load_dataset("titanic")

# Histplot: male vs female
ax = sns.histplot(data=titanic, x="sex", hue="sex", shrink=0.6, legend=False)

# Bar-er upor count dekhano
for container in ax.containers:
    ax.bar_label(container)

plt.title("Titanic: Male vs Female")
plt.xlabel("Sex")
plt.ylabel("Count")
plt.show()
```
